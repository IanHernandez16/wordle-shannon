```python
import urllib.request, os
url = "https://raw.githubusercontent.com/3b1b/videos/master/_2022/wordle/data/possible_words.txt"
if os.path.exists("ans.txt"):
    os.remove("ans.txt")
urllib.request.urlretrieve(url, "ans.txt")
words = [w.strip() for w in open("ans.txt") if len(w.strip()) == 5]
print(len(words), words[:5])
```

```python
import numpy as np, math, random, sys
W=[w.strip().lower() for w in open('ans.txt') if len(w.strip())==5]
W=sorted(set(W)); M=len(W); print("M =",M)
def pat(g,t):
    r=[0]*5; cnt={}
    for i in range(5):
        if g[i]==t[i]: r[i]=2
        else: cnt[t[i]]=cnt.get(t[i],0)+1
    for i in range(5):
        if r[i]==0 and cnt.get(g[i],0)>0: r[i]=1; cnt[g[i]]-=1
    return sum(r[i]*3**i for i in range(5))
P=np.zeros((M,M),dtype=np.uint8)
for i,g in enumerate(W):
    for j,t in enumerate(W): P[i,j]=pat(g,t)
np.save('P.npy',P)
def entropy_of(g,cand):
    c=np.bincount(P[g,cand],minlength=243); c=c[c>0]; p=c/c.sum()
    return -(p*np.log2(p)).sum()
allidx=np.arange(M)
H=np.array([entropy_of(g,allidx) for g in range(M)])
order=np.argsort(-H)
print("H(X)=log2 M =",math.log2(M))
print("Top 5 first words:",[(W[i],round(H[i],3)) for i in order[:5]])
first=order[0]
def play_shannon(t,first=first):
    cand=allidx.copy(); g=first; n=0
    while True:
        n+=1
        if g==t: return n
        cand=cand[P[g,cand]==P[g,t]]
        if len(cand)==1: g=cand[0]; continue
        cs=set(cand.tolist())
        best=None;bh=-1
        for k in (cand if len(cand)>60 else allidx):
            h=entropy_of(k,cand)+(1e-9 if k in cs else 0)
            if h>bh: bh=h;best=k
        g=best
        if n>=15: return n
def play_random(t,rng,fixed=None):
    cand=allidx.copy(); n=0; g=fixed if fixed is not None else rng.choice(cand)
    while True:
        n+=1
        if g==t: return n
        cand=cand[P[g,cand]==P[g,t]]
        g=rng.choice(cand)
        if n>=15: return n
rng=np.random.default_rng(1)
def stats(res):
    res=np.array(res); return res.mean(), (res<=6).mean()*100, res.max(), np.bincount(res)
rs=[play_shannon(t) for t in range(M)]
print("SHANNON",stats(rs))
rr=[play_random(t,rng) for t in range(M)]
print("RANDOM",stats(rr))
fx=W.index('crane') if 'crane' in W else None
if fx is not None:
    rf=[play_random(t,rng,fx) for t in range(M)]
    print("FIXED crane + random",stats(rf))
# worked example
tgt=W.index('shard') if 'shard' in W else 0
cand=allidx.copy(); g=first
print("TARGET",W[tgt])
for step in range(1,4):
    pt=P[g,tgt]; Hc=-sum((c/len(cand))*math.log2(c/len(cand)) for c in np.bincount(P[g,cand],minlength=243) if c)
    new=cand[P[g,cand]==pt]
    sym=''.join('GYB'[ [2,1,0].index(v) ] if False else '⬛🟨🟩'[v] for v in [(pt//3**i)%3 for i in range(5)])
    print(step,W[g],sym,"before",len(cand),"after",len(new),"H(X)=%.3f"%math.log2(len(cand)),"E[I]=%.3f"%Hc,"I_real=%.3f"%(math.log2(len(cand))-math.log2(len(new))),"H(X|Y)exp=%.3f"%(math.log2(len(cand))-Hc))
    cand=new
    if len(cand)==1 or W[g]==W[tgt]: break
    cs=set(cand.tolist()); best=None;bh=-1
    for k in (cand if len(cand)>60 else allidx):
        h=entropy_of(k,cand)+(1e-9 if k in cs else 0)
        if h>bh: bh=h;best=k
    g=best
```

## Análisis adicionales para la versión final
Estas celdas reproducen las cifras nuevas de la infografía: referencias aleatorias con 30 semillas, ejemplo SHARD con candidatas, palabras que necesitan 6 intentos, cota de información por pista y gráfico de distribución.

```python
# 1) Cota de información por pista y uso de patrones de RAISE
print("Máx. bits por pista: log2(243) =", round(math.log2(243),3))
print("Pistas mínimas en promedio (cota):", round(math.log2(M)/math.log2(243),2))
c=np.bincount(P[first],minlength=243)
print(W[first],"usa",(c>0).sum(),"de 243 patrones; grupo más grande:",c.max())
```

```python
# 2) Ejemplo SHARD con lista de candidatas y colores
def sym(pt): return ''.join('⬛🟨🟩'[(pt//3**i)%3] for i in range(5))
def choose(cand):
    cs=set(cand.tolist()); best=None; bh=-1
    for k in (cand if len(cand)>60 else allidx):
        h=entropy_of(k,cand)+(1e-9 if k in cs else 0)
        if h>bh: bh=h; best=k
    return best
t=W.index('shard'); cand=allidx.copy(); g=first
while True:
    pt=P[g,t]; new=cand[P[g,cand]==pt]
    print(W[g].upper(), sym(pt), len(cand),'->',len(new), [W[i] for i in new] if 1<len(new)<=25 else '')
    if g==t: break
    cand=new; g=cand[0] if len(cand)==1 else choose(cand)
# ¿Por qué DUCHY? entropía de cada candidata vs DUCHY con las 5 restantes
c5=np.array([W.index(w) for w in ['scarf','scary','shard','shark','swarm']])
for w in ['scarf','scary','shard','shark','swarm','duchy']:
    k=W.index(w); cnt=np.bincount(P[k,c5],minlength=243); print(w, round(entropy_of(k,c5),3), sorted(cnt[cnt>0].tolist(),reverse=True))
```

```python
# 3) Palabras que necesitan 6 intentos con la estrategia de Shannon
def path(t):
    cand=allidx.copy(); g=first; p=[]
    while True:
        p.append(W[g])
        if g==t: return p
        cand=cand[P[g,cand]==P[g,t]]; g=cand[0] if len(cand)==1 else choose(cand)
print([(W[t],path(t)) for t in range(M) if rs[t]>=6])
```

```python
# 4) Referencias aleatorias promediadas sobre 30 semillas
fx=W.index('crane'); out={}
for name,fixed in [('CRANE + azar',fx),('Azar',None)]:
    means=[];wins=[];mx=[];dist=np.zeros(20)
    for s in range(30):
        rng=np.random.default_rng(s)
        r=np.array([play_random(t,rng,fixed) for t in range(M)])
        means.append(r.mean()); wins.append((r<=6).mean()*100); mx.append(r.max()); dist[:r.max()+1]+=np.bincount(r)
    out[name]=dist/30
    print(f"{name}: promedio {np.mean(means):.2f} ± {np.std(means):.2f} | ≤6: {np.mean(wins):.1f}% ({min(wins):.1f}-{max(wins):.1f}) | máx {max(mx)}")
```

```python
# 5) Gráfico de distribución de intentos
import matplotlib.pyplot as plt
sh=np.zeros(20); b=np.bincount(rs); sh[:len(b)]=b
xs=np.arange(1,10); w=0.27
fig,ax=plt.subplots(figsize=(7,3.5))
for k,(d,col,lab) in enumerate([(sh,'#2a78d6','Shannon (RAISE)'),(out['CRANE + azar'],'#eb6834','CRANE + candidata al azar'),(out['Azar'],'#1baf7a','Candidata al azar')]):
    ax.bar(xs+(k-1)*w,d[1:10]/M*100,width=w-0.03,color=col,label=lab)
ax.axvline(6.5,color='gray',ls='--'); ax.set_xticks(xs)
ax.set_xlabel('Intentos necesarios'); ax.set_ylabel('% de partidas'); ax.legend(); plt.tight_layout(); plt.show()
```
